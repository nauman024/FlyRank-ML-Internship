## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
import numpy as np

# 1. Load starter dataset directly from your public repository
repo_url = "https://raw.githubusercontent.com/nauman024/FlyRank-ML-Internship/main/data/raw/content_refresh_anonymized.csv"

try:
    df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
except FileNotFoundError:
    df = pd.read_csv(repo_url)

# 2. Define Unit of Analysis
print("==================================================")
print("UNIT OF ANALYSIS: One row = One Page / URL")
print(f"Total pages analyzed: {len(df):,}")
print("==================================================\n")

# 3. Construct Proxy Target Column for Lane 1 framing
# Target Definition: 1 if page is trending down AND has position rank decay (> 15)
df['is_underperforming'] = np.where((df['trend_direction'] == 'down') & (df['avg_position'] > 15), 1, 0)

# 4. Display slice showing Unit of Analysis + Target
feature_slice = ['url_hash', 'content_type', 'content_age_days', 'avg_position', 'ctr', 'trend_direction', 'is_underperforming']

# Fallback if specific columns differ slightly in naming
available_cols = [c for c in feature_slice if c in df.columns]
display_df = df[available_cols].head(10)

print("Sample DataFrame with Target Sketch:")
display(display_df)

# 5. Target Class Distribution
target_counts = df['is_underperforming'].value_counts(normalize=True) * 100
print("\nTarget Class Distribution:")
print(f"Healthy / Stable (0): {target_counts.get(0, 0):.2f}%")
print(f"At Risk / Underperforming (1): {target_counts.get(1, 0):.2f}%")
```

## 5. Why ML beats a fixed rule here

Fixed heuristics (e.g., "if position > 20 and traffic dropped 10%") fail because search decay is non-linear and context-dependent. A newly published page naturally behaves differently than an aged cornerstone page. Machine learning models capture high-dimensional feature interactions—such as content age, position variance, and click-through rates across query clusters—that simple nested if-else rules miss.
