## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# -------------------------------------------------------------------
# 1. Load the starter dataset directly from your GitHub repository
# Replace 'YOUR_GITHUB_USERNAME' and 'YOUR_REPO_NAME' with your actual GitHub details!
# -------------------------------------------------------------------
repo_url = "https://raw.githubusercontent.com/nauman024/FlyRank-ML-Internship/main/data/raw/content_refresh_anonymized.csv"

try:
    df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
except FileNotFoundError:
    df = pd.read_csv(repo_url)

# -------------------------------------------------------------------
# 2. Number 1: Total pages vs decaying/down pages
# -------------------------------------------------------------------
total_pages = len(df)
down_pages = len(df[df['trend_direction'] == 'down'])
down_pct = (down_pages / total_pages) * 100

# -------------------------------------------------------------------
# 3. Number 2: Zero-impression rate (inactive pages over 90 days)
# -------------------------------------------------------------------
zero_impressions = len(df[df['impressions_90d'] == 0])
zero_imp_pct = (zero_impressions / total_pages) * 100

# -------------------------------------------------------------------
# 4. Number 3: Average CTR for top position tier vs lower positions
# -------------------------------------------------------------------
top_tier_ctr = df[df['avg_position'] <= 10]['ctr'].mean()
lower_tier_ctr = df[df['avg_position'] > 10]['ctr'].mean()

# -------------------------------------------------------------------
# 5. Output Results
# -------------------------------------------------------------------
print(f"Dataset successfully loaded! Total pages: {total_pages:,}\n")
print(f"Number 1: {down_pages:,} out of {total_pages:,} pages ({down_pct:.1f}%) are currently trending down.")
print(f"Number 2: {zero_impressions:,} pages ({zero_imp_pct:.1f}%) have 0 impressions over 90 days.")
print(f"Number 3: Top-tier pages (avg_position <= 10) have an average CTR of {top_tier_ctr:.2%}, compared to {lower_tier_ctr:.2%} for lower positions.")
```
